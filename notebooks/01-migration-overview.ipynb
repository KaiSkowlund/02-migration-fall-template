# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?

Many animals migrate to follow resources and suitable conditions as the seasons change. Many birds, like the Veery, fly north in the spring to breed. Summer brings higher abundance of insects, daylight, and better temperature conditions. In the fall, they return south before the temperature drops. Elk and mule deer migrate to higher or lower elevation to find more suitable conditions based on the time of year.

2. How might climate change affect animal migrations?

Climate change impacts conditions that animals are looking for when they migrate. Temperature and precipitation changes may impact resource availability in areas animals migrate to or from. Often migration for birds can be based on how long days are, and when their migration is timed with that, and there has been a shift in resource availability in the area which they migrate to, they may mis the conditions that serve the purpose for their migration. 

3. Do you notice any animal migrations where you live?

The two most notable migration I notice living in Boulder, are Elk and Canadian geese. In the fall and winter, elk are more prevalent in the lower elevation areas, especially near Estes Park. In the summer, they push up to higher elevations to escape the heat. In the spring and fall, the canadian geese take over campus during their migration and make their presence known.
