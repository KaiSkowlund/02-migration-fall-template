The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

### Reflect and Respond #1: Reflect on what you know about migration. You could consider:
1. What are some reasons that animals might migrate?

Animals migrate to follow resources and suitable conditions as the seasons change. Many songbirds, like the Veery, fly north in the spring to breed, because summers in the north offer an abundance of insects, more daylight, and more optimal temperatures for raising young. In the fall, they return to South America before winter hits in the north and food becomes scarce. Not all migrations are over long distances. Elk and mule deer migrate based on elevation. In the hot summer months, they move up into the high country to follow fresh vegetation as the snow melts, then in the winter move down in elevation when deep snow covers their food. (National Park Service)

2. How might climate change affect animal migrations?

Climate change impacts conditions that animals are looking for when they migrate. Many birds use cues like day length, which is not impacted at all by climate change, but climate change is impacting the conditions that they are traveling long distances for. Temperature driven events, such as spring green-up and insect emergence, are happening earlier. When birds fall out of sync with these events, they can arrive after the peak in food availability. This is called phenological mismatch. Saino et al. (2011) studied European migratory birds and found that long-distance migrants had not advanced their spring arrival as much as spring conditions had advanced. Species with a larger mismatch also showed greater population declines. Climate change can also shrink or shift the habitat migrants depend on. Audubon’s Survival by Degrees report found that 64% of North American bird species. 389 of 604 are at risk of extinction from climate change. In the arid west, drought, extreme heat, longer fire seasons, and earlier snow melt, are all reducing the habitat that migratory birds. This may result in certain birds, such as the sandhill crane, shifting their migration further north, where there are more optimal conditions.

3. Do you notice any animal migrations where you live?

The two most notable migration I notice living in Boulder, are Elk and Canada geese. In the fall and winter, elk are more prevalent in the lower elevation areas, especially near Estes Park. In the summer, they push up to higher elevations to escape the heat. The NPS notes that elk have been shifting further east toward the lower Front Range since the early 2000s.In the spring and fall, large flocks of Canada geese take over campus and fly overhead as they pass through. Some geese along the Front Range now stay year-round, so the flocks I see are likely a mix of migrants and resident birds.
